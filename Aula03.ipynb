Exercício 1 — Referências
Execute:


````python
a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)

print("b:", b)

print("id(a):", id(a))

print("id(b):", id(b))
````

Responda:

Qual será a saída?

a e b representam o mesmo objeto?

Por quê?

O que aconteceria com b = a.copy()?

**Resposta**
1. A saída será:


```
a: [10, 20, 30, 40]
b: [10, 20, 30, 40]
id(a): mesmo valor de id(b)
id(b): mesmo valor de id(a)
```
2. Sim, a e b representam o mesmo objeto.
3. Porque b = a faz b receber uma referência para o mesmo objeto que a.
4. Com b = a.copy(), b seria uma cópia independente de a.
5. Alterações em b não alterariam a.


Exercício 2 — Construindo uma cadeia
Crie:

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

Conecte:

A → B → C → None

Depois percorra a estrutura e produza:

A

B

C

**Resposta**

````python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3
n3.proximo = None

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
````

Exercício 3 — Depuração

Analise:

````python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
````

Perguntas:

Qual é o problema?

Por que o programa não termina?

Qual linha deve ser acrescentada?

Qual será a saída depois da correção?

**Resposta**
1. O problema é que atual nunca é atualizado dentro do while.

2. O programa não termina porque atual permanece apontando para n1.

3. Deve ser acrescentada a linha:


```
atual = atual.proximo
```
4. A saída será:

```
10
20
30
```
